```python
# create student dictionary & print whole dictionary
student = {
"name": "Rahul",
"age": 22,
"course": "Python",
"city": "Bangalore",
"marks": 88
}

print(student)
print(type(student))
```

```python
#Print the student's name
print(student["name"])
```

```python
#Print their course
print(student["course"])
```

```python
#Print all keys
print(student.keys())
```

```python
#Print all values
print(student.values())
```

```python
#Print all key-value pairs
print(student.items())
```

```python
student["marks"] = 92
print(student)
```

```python
#Add "email"
student["email"]= "rahul@gmail.com"
print(student)
```

```python
#Add "phone"
student["phone"] = "123456789"
print(student)
```

```python
#Remove "city"
student.pop("city")
print(student)
```

```python
#Use get() to retrieve "name"
print(student.get("name"))
```

```python
#Create a copy of the dictionary
student1= student.copy()
print(student1)
```
