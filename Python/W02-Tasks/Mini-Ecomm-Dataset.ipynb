```python
#Create a Mini E-Commerce Dataset

products = [
    {
        "name": "Headphones",
        "price": 15000,
        "brand": "Sony"
    },
    {
        "name": "Smartwatch",
        "price": 25000,
        "brand": "Garmin"
    },
    {
        "name": "Monitor",
        "price": 35000,
        "brand": "LG"
    }
]
```

```python
#Print all products without using a loop
print(products)
```

```python
print(type(products))  # Verify the data type of products
```

```python
#Print first product
print(products[0])
```

```python
#Print second product's price
print(products[1]["price"])
```

```python
#Print third product's brand
print(products[2]["brand"])
```

```python
#Change first product's price
products[0]["price"] = 45000
print(products[0])
```

```python
#Add "rating" to the second product
products[1]["rating"] = 4.5
print(products[1])
```

```python
print(products)
```
