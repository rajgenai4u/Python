```python
#Create list cart
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

```python
#Display all products
print(cart)
```

```python
#Access first and last products 
print(cart[0])

print(cart[-1])
```

```python
#Add "Webcam"
cart.append("Webcam")
print(cart)
```

```python
#Insert "USB Hub" at index 2
cart[2] = "USB Hub"
print(cart)
```

```python
#Remove "Mouse"
cart.remove("Mouse")
print(cart)
```

```python
#Remove the last item using pop()
cart.pop()
print(cart)
```

```python
#Find the index of "Monitor"
print(cart.index("Monitor"))
```

```python
#Count occurrences of "Laptop"
print(cart.count("Laptop"))
```

```python
#Create a copy of the cart
cart1=cart.copy()
#creates separate list
print(cart1)
```

```python
cart2= cart1 #Creates another reference to the same list
print(cart2)
```

```python
#Reverse the cart
#revcart=cart.reverse()
revcart = cart[::-1]
print(revcart)
```

```python
#Sort the products alphabetically
sortcart=sorted(cart)
#creates a new sorted list
print(sortcart)
```

```python
#sort the original list
#print(cart)
sortcart1=cart.sort()
#print(sortcart1)
print(cart)
```
